# Logical Elements

```python
# Initialize driver and maximize window
driver = webdriver.Edge()
driver.maximize_window()

# Navigate to Google
URL = "https://www.google.com"
driver.get(URL)

# Wait a moment for the page to fully load (consider using explicit waits if necessary)
time.sleep(2)

# Locate the search bar and send input
search_bar_xpath = '//*[@name="q"]'  # Alternate and commonly used attribute for Google's search bar
search_bar = driver.find_element(By.XPATH, search_bar_xpath)
search_bar.send_keys("Machine Learning")
time.sleep(1)
search_bar.send_keys(Keys.ENTER)

# Use explicit wait for the search results
wait = WebDriverWait(driver, 10)
try:
    # You may need to adjust the XPath here based on the current layout
    link = wait.until(EC.element_to_be_clickable(
        (By.XPATH, '//*[@id="rso"]//h3[1]')
    ))
    link.click()
except Exception as e:
    print("Error locating or clicking the search result element:", e)
finally:
    # Optional: close the driver after a short delay
    time.sleep(5)
    driver.quit()
```

# Working with MultiSelect

# Basic Scrolling

```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
```

```python
driver = webdriver.Chrome()
driver.maximize_window()

url = 'https://en.wikipedia.org/wiki/Machine_learning'
driver.get(url)
time.sleep(2)
# driver.quit()

# Scrolling by element
ai_xpath = '//*[@id="Artificial_intelligence"]'
ai_subtopic = driver.find_element(By.XPATH, ai_xpath)
driver.execute_script("arguments[0].scrollIntoView();",ai_subtopic)

# Scrolling Vertically
driver.execute_script('window.scrollBy(0,3000);')
time.sleep(2)
driver.execute_script('window.scrollBy(0,-1500);')

# Scrolling by page hight
driver.execute_script("window.scrollTo(0,document.body.scrollHeight);")
time.sleep(2)
driver.execute_script("window.scrollTo(0,-document.body.scrollHeight);")
```

```python
import time
from selenium import webdriver
from selenium.webdriver.common.by import By

driver = webdriver.Chrome()
driver.maximize_window()

url = "https://www.w3schools.com/html/tryit.asp?filename=tryhtml_iframe_target"
driver.get(url)
time.sleep(2)

# identify iframe element and switch context
iframe_element = driver.find_element(By.XPATH, '//*[@id="iframeResult"]')
driver.switch_to.frame(iframe_element)

link = driver.find_element(By.XPATH, '/html/body/p[1]/a')
link.click()
time.sleep(2)

driver.switch_to.default_content()
driver.quit()
```

```python
driver = webdriver.Chrome()
driver.maximize_window()

url = "https://github.com/login"
driver.get(url)
time.sleep(2)

username_field = driver.find_element(By.ID,"login_field")
username_field.send_keys("Aman")

password_field = driver.find_element(By.ID,"password")
password_field.send_keys("pass123")
time.sleep(2)

login_button = driver.find_element(By.NAME,"commit")
login_button.click()
```

```python
# POM : Page Object Model
driver = webdriver.Chrome()
driver.maximize_window()

url = "https://github.com/login"
driver.get(url)
time.sleep(2)

class LoginPage:
	def __init__(self, driver):
		self.driver = driver
		self.username = (By.ID, "login_field")
		self.password = (By.ID, "password")
		self.login_button = (By.NAME, "commit")


	def login(self, username, password):
		self.driver.find_element(*self.username).send_keys(username)
		self.driver.find_element(*self.password).send_keys(password)
		time.sleep(1)
		self.driver.find_element(*self.login_button).click()


login_page = LoginPage(driver)
login_page.login("misbah", "pass123")

time.sleep(2)
driver.quit()
```
